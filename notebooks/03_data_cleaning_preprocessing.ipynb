##### 4. Initial Cleaning Decisions

```python
df_clean = df.copy()

# Remove exported identifier
df_clean = df_clean.drop(
    columns=["Unnamed: 0"],
    errors="ignore"
)

# Remove observations without the target
df_clean = df_clean.dropna(
    subset=["Price"]
)

print("Shape after removing missing target:", df_clean.shape)

# Check duplicates after identifier removal
duplicate_count = df_clean.duplicated().sum()

print("Duplicates after identifier removal:", duplicate_count)

# Remove duplicate observations
df_clean = df_clean.drop_duplicates()

print("Final shape after initial cleaning:", df_clean.shape)
```

##### 5. Categorical Value Consistency Check

```python
categorical_columns = [
    "Brand",
    "Model",
    "Gear",
    "Fuel Type",
    "Town",
    "Leasing",
    "Condition",
    "AIR CONDITION",
    "POWER STEERING",
    "POWER MIRROR",
    "POWER WINDOW"
]

for col in categorical_columns:
    print(f"\n{col}")
    print("Unique values:", df_clean[col].nunique())

    if df_clean[col].nunique() <= 10:
        print(df_clean[col].value_counts(dropna=False))
```

```python
print("Leading/trailing whitespace check:")

for col in categorical_columns:
    whitespace_count = (
        df_clean[col]
        .dropna()
        .astype(str)
        .apply(lambda value: value != value.strip())
        .sum()
    )

    print(f"{col}: {whitespace_count}")
```

##### 6. Numerical Value Validation and Outliers

```python
numeric_columns = [
    "YOM",
    "Engine (cc)",
    "Mileage(KM)",
    "Price"
]

df_clean[numeric_columns].describe()
```

```python
print("YOM below 1900:",
      (df_clean["YOM"] < 1900).sum())

print("YOM above 2025:",
      (df_clean["YOM"] > 2025).sum())

print("Engine <= 0:",
      (df_clean["Engine (cc)"] <= 0).sum())

print("Mileage < 0:",
      (df_clean["Mileage(KM)"] < 0).sum())

print("Price <= 0:",
      (df_clean["Price"] <= 0).sum())
```

```python
def count_iqr_outliers(series):
    series = series.dropna()

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    count = (
        (series < lower_bound) |
        (series > upper_bound)
    ).sum()

    return count, lower_bound, upper_bound


for col in numeric_columns:
    count, lower, upper = count_iqr_outliers(df_clean[col])

    print(
        f"{col}: {count} statistical outliers "
        f"(bounds: {lower:.2f} to {upper:.2f})"
    )
```

##### 13. Final Feature Groups

```python
numeric_features = [
    "Engine (cc)",
    "Vehicle_Age",
    "Feature_Count"
]

categorical_features = [
    "Brand",
    "Model",
    "Gear",
    "Fuel Type",
    "Town",
    "Leasing",
    "Condition",
    "AIR CONDITION",
    "POWER STEERING",
    "POWER MIRROR",
    "POWER WINDOW",
    "Listing_Month"
]

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

print(
    "\nTotal selected features:",
    len(numeric_features) + len(categorical_features)
)
```

##### 14. Build Final Preprocessing Pipelines

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

```python
#create the numerical pipeline for linear models
numeric_linear_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
#create the numerical pipeline for tree models
numeric_tree_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])
```

```python
categorical_pipeline = Pipeline([
    ("encoder", OneHotEncoder(
        handle_unknown="infrequent_if_exist",
        min_frequency=10
    ))
])
```

##### 15. Create the final preprocessors

```python
linear_preprocessor = ColumnTransformer([
    ("num", numeric_linear_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
tree_preprocessor = ColumnTransformer([
    ("num", numeric_tree_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

##### 16. Validate the preprocessing

```python
X_train_linear_processed = linear_preprocessor.fit_transform(
    X_train_engineered
)

X_test_linear_processed = linear_preprocessor.transform(
    X_test_engineered
)

X_train_tree_processed = tree_preprocessor.fit_transform(
    X_train_engineered
)

X_test_tree_processed = tree_preprocessor.transform(
    X_test_engineered
)
```

```python
print("Linear preprocessing")
print("Train shape:", X_train_linear_processed.shape)
print("Test shape:", X_test_linear_processed.shape)

print("\nTree preprocessing")
print("Train shape:", X_train_tree_processed.shape)
print("Test shape:", X_test_tree_processed.shape)
```

```python
from scipy import sparse
import numpy as np

def count_missing_values(X):
    if sparse.issparse(X):
        return np.isnan(X.data).sum()
    return np.isnan(X).sum()

print(
    "Linear train missing:",
    count_missing_values(X_train_linear_processed)
)

print(
    "Linear test missing:",
    count_missing_values(X_test_linear_processed)
)

print(
    "Tree train missing:",
    count_missing_values(X_train_tree_processed)
)

print(
    "Tree test missing:",
    count_missing_values(X_test_tree_processed)
)
```
