####  Data Cleaning and Preprocessing

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_PATH = Path("../data/raw/car_price_dataset.csv")

df = pd.read_csv(DATA_PATH)

df.head()
```

 ##### 1. Dataset Overview

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

##### 1. Missing Value Check


```python
missing_values = df.isna().sum()

print(missing_values[missing_values > 0])
```

##### 3. Duplicate Check

```python
print("Exact duplicates before removing identifier:")
print(df.duplicated().sum())

print("\nUnnamed: 0 sample:")
print(df["Unnamed: 0"].head())

print("\nUnique Unnamed: 0 values:")
print(df["Unnamed: 0"].nunique())
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

duplicate_rows
```

##### 4. Initial Cleaning Decisions

```python
df_clean = df.copy()

# Remove exported identifier
df_clean = df_clean.drop(
    columns=["Unnamed: 0"],
    errors="ignore"
)

# Remove observations without the target
df_clean = df_clean.dropna(
    subset=["Price"]
)

print("Shape after removing missing target:", df_clean.shape)

# Check duplicates after identifier removal
duplicate_count = df_clean.duplicated().sum()

print("Duplicates after identifier removal:", duplicate_count)

# Remove duplicate observations
df_clean = df_clean.drop_duplicates()

print("Final shape after initial cleaning:", df_clean.shape)
```

##### 5. Categorical Value Consistency Check

```python
categorical_columns = [
    "Brand",
    "Model",
    "Gear",
    "Fuel Type",
    "Town",
    "Leasing",
    "Condition",
    "AIR CONDITION",
    "POWER STEERING",
    "POWER MIRROR",
    "POWER WINDOW"
]

for col in categorical_columns:
    print(f"\n{col}")
    print("Unique values:", df_clean[col].nunique())

    if df_clean[col].nunique() <= 10:
        print(df_clean[col].value_counts(dropna=False))
```

```python
print("Leading/trailing whitespace check:")

for col in categorical_columns:
    whitespace_count = (
        df_clean[col]
        .dropna()
        .astype(str)
        .apply(lambda value: value != value.strip())
        .sum()
    )

    print(f"{col}: {whitespace_count}")
```

##### 6. Numerical Value Validation and Outliers

```python
numeric_columns = [
    "YOM",
    "Engine (cc)",
    "Mileage(KM)",
    "Price"
]

df_clean[numeric_columns].describe()
```

```python
print("YOM below 1900:",
      (df_clean["YOM"] < 1900).sum())

print("YOM above 2025:",
      (df_clean["YOM"] > 2025).sum())

print("Engine <= 0:",
      (df_clean["Engine (cc)"] <= 0).sum())

print("Mileage < 0:",
      (df_clean["Mileage(KM)"] < 0).sum())

print("Price <= 0:",
      (df_clean["Price"] <= 0).sum())
```

```python
def count_iqr_outliers(series):
    series = series.dropna()

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    count = (
        (series < lower_bound) |
        (series > upper_bound)
    ).sum()

    return count, lower_bound, upper_bound


for col in numeric_columns:
    count, lower, upper = count_iqr_outliers(df_clean[col])

    print(
        f"{col}: {count} statistical outliers "
        f"(bounds: {lower:.2f} to {upper:.2f})"
    )
```

##### 7. YOM and Mileage Relationship

```python
df_clean[["YOM", "Mileage(KM)"]].corr()
```

```python
mileage_per_year = (
    df_clean
    .dropna(subset=["YOM", "Mileage(KM)"])
    .groupby("YOM")["Mileage(KM)"]
    .nunique()
)

print(
    "Maximum number of mileage values for one YOM:",
    mileage_per_year.max()
)

print(
    "Years with more than one mileage value:",
    (mileage_per_year > 1).sum()
)
```

##### 8. Date Validation

```python
df_clean["Date"] = pd.to_datetime(
    df_clean["Date"],
    errors="coerce"
)

print(
    "Invalid parsed dates:",
    df_clean["Date"].isna().sum()
)

print(
    "Earliest date:",
    df_clean["Date"].min()
)

print(
    "Latest date:",
    df_clean["Date"].max()
)
```

##### 9. Final Cleaned Dataset Check

```python
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isna().sum()[df_clean.isna().sum() > 0])

print("\nCleaned columns:")
print(df_clean.columns.tolist())
```

##### 10. Feature–Target Separation and Shared Train/Test Split

```python
from sklearn.model_selection import train_test_split

# Separate input features and target
X = df_clean.drop(columns=["Price"])
y = df_clean["Price"]

# Create shared train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print(
    "Training percentage:",
    round(len(X_train) / len(X) * 100, 2),
    "%"
)

print(
    "Testing percentage:",
    round(len(X_test) / len(X) * 100, 2),
    "%"
)

print("\nFeatures passed to Feature Engineering:")
print(X_train.columns.tolist())
```
